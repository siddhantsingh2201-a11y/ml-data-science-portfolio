```python
"""
MMM or Market Mix Modeling is the statistically analysis strategy.
It is a top-down statistical analysis (usually Regression) that looks at historical data,
To quantify the impact of various marketing channels and external factors on sales.

"""
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
```

```python
## $$Sales = \beta_0 + \beta_1(TV\_Spend) + \beta_2(Social\_Spend) + \beta_3(Seasonality) + \dots + Error$$
## Where $\beta$ (Beta) tells you the "lift" or "ROI" of that specific channel. 
## If $\beta_{TV} = 5$, it means for every $1 spent on TV, you got $5 in sales.
```

```python
# Synthetic Data
np.random.seed(42)
weeks = 104
data = pd.DataFrame({'Week' : pd.date_range(start= '2025-01-01', periods=weeks , freq='W')})
```

```python
data['TV_Spend'] = np.random.normal(10000, 2000, weeks)
data['Social_Spend'] = np.random.normal(5000, 1500, weeks)
data['Radio_Spend'] = np.random.normal(3000, 800, weeks)
```

```python
# 2. Define the "True" ROI (The hidden truth we want to find)
# TV ROI = 3.5 (Returns $3.5 for every $1)
# Social ROI = 5.0 (Returns $5.0 for every $1)
# Radio ROI = 1.2 (Returns $1.2 for every $1)
```

```python
true_roi = {'TV' : 3.5, 'Social' : 5.0, 'Radio' : 1.2}
```

```python
base_sales = 50000
```

```python
data['Sales'] = (base_sales + (data['TV_Spend'] * true_roi['TV']) + 
               (data['Social_Spend'] * true_roi['Social']) + (data['Radio_Spend'] * true_roi['Radio']) + np.random.normal(0,5000, weeks))
```
